```python
# --- Carga segura de FastSAM en PyTorch 2.6+ con checkpoints de Ultralytics ---

import inspect, torch, torch.nn as nn
from fastsam import FastSAM, FastSAMPrompt

def _allowlist_torch_and_ultralytics():
    # 1) Todas las clases de torch.nn.Module (Conv2d, BatchNorm2d, ReLU, Sequential, etc.)
    safe = []
    for name, obj in vars(nn).items():
        if inspect.isclass(obj) and issubclass(obj, nn.Module):
            safe.append(obj)
    try:
        torch.serialization.add_safe_globals(safe)
    except Exception:
        pass

    # 2) Todas las clases nn.Module de ultralytics.nn.modules (incluye Concat, C2f, Conv, etc.)
    try:
        import ultralytics.nn.modules as um
        safe_ultra = []
        for name, obj in vars(um).items():
            if inspect.isclass(obj) and issubclass(obj, nn.Module):
                safe_ultra.append(obj)
        if safe_ultra:
            torch.serialization.add_safe_globals(safe_ultra)
    except Exception:
        pass

    # 3) Modelos/tareas de Ultralytics (p.ej. SegmentationModel)
    try:
        import ultralytics.nn.tasks as ut
        safe_tasks = []
        for name, obj in vars(ut).items():
            if inspect.isclass(obj) and issubclass(obj, nn.Module):
                safe_tasks.append(obj)
        if safe_tasks:
            torch.serialization.add_safe_globals(safe_tasks)
    except Exception:
        pass


def build_fastsam_safe(ckpt_path, try_insecure_fallback=True):
    """
    Construye FastSAM registrando clases para weights_only=True.
    Si aún falla y try_insecure_fallback=True, fuerza weights_only=False
    SOLO durante la construcción (confía en el origen del .pt).
    """
    _allowlist_torch_and_ultralytics()

    # Intento 1: con weights_only=True (por defecto en PyTorch 2.6+)
    try:
        return FastSAM(ckpt_path)
    except Exception as e1:
        last_err = e1

    # Intento 2 (opcional): monkeypatch temporal para forzar weights_only=False
    if try_insecure_fallback:
        orig_torch_load = torch.load

        def patched_load(*args, **kwargs):
            kwargs["weights_only"] = False
            return orig_torch_load(*args, **kwargs)

        torch.load = patched_load
        try:
            model = build_fastsam_safe(ckpt)
            return model
        finally:
            # Restaurar torch.load sí o sí
            torch.load = orig_torch_load

    # Si nada funcionó, relanza el error original para debugear
    raise last_err
```

```python
# ---------- 1) (Opcional) Construcción de máscaras desde LabelMe-like JSON ----------
# Si ya tienes image_paths y mask_paths creados por tu script, puedes saltarte esta celda.
# De lo contrario, ajusta labeled_dir y ejecuta para generar masks PNG 0/255 en generated_masks.

labeled_dir = "/content/drive/MyDrive/Colab Notebooks/SolDef_AI/Labeled"
mask_output_dir = os.path.join(labeled_dir, "generated_masks")
os.makedirs(mask_output_dir, exist_ok=True)

def scale_and_clamp_polygon(points, ann_w, ann_h, img_w, img_h):
    sx = img_w / float(ann_w)
    sy = img_h / float(ann_h)
    out = []
    for (x, y) in points:
        xi = int(round(max(0, min(img_w - 1, x * sx))))
        yi = int(round(max(0, min(img_h - 1, y * sy))))
        out.append((xi, yi))
    return out

def build_image_and_mask_lists():
    image_paths, mask_paths = [], []
    if not os.path.isdir(labeled_dir):
        print(f"⚠️ No existe {labeled_dir}. Si ya tienes listas, ignora esta función.")
        return image_paths, mask_paths

    json_files = sorted([f for f in os.listdir(labeled_dir) if f.lower().endswith(".json")])
    print("JSONs encontrados:", len(json_files))

    for jf in json_files:
        jp = os.path.join(labeled_dir, jf)
        try:
            with open(jp, "r") as f:
                data = json.load(f)
        except Exception as e:
            print(f"⚠️ No se pudo leer {jp}: {e}")
            continue

        image_filename = data.get("imagePath")
        if not image_filename:
            print(f"⚠️ {jp} sin imagePath, se salta.")
            continue

        ip = os.path.join(labeled_dir, image_filename)
        if not os.path.exists(ip):
            print(f"⚠️ Imagen no existe: {ip}, se salta.")
            continue

        img = Image.open(ip).convert("RGB")
        img_w, img_h = img.size
        ann_w = data.get("imageWidth", img_w)
        ann_h = data.get("imageHeight", img_h)

        mask = Image.new("L", (img_w, img_h), 0)
        draw = ImageDraw.Draw(mask)
        n_polys = 0
        for shape in data.get("shapes", []):
            pts = shape.get("points", [])
            if not pts:
                continue
            st = shape.get("shape_type", "polygon")
            if st not in ("polygon", "polyline", "linestrip", "rectangle"):
                continue
            if st == "rectangle" and len(pts) == 2:
                (x0, y0), (x1, y1) = pts
                pts = [(x0, y0), (x1, y0), (x1, y1), (x0, y1)]
            scaled = scale_and_clamp_polygon(pts, ann_w, ann_h, img_w, img_h)
            if len(scaled) >= 3:
                draw.polygon(scaled, outline=255, fill=255)
                n_polys += 1

        if n_polys == 0:
            print(f"ℹ️ Sin polígonos válidos en {jp}, se salta.")
            continue

        base = os.path.splitext(os.path.basename(image_filename))[0]
        mp = os.path.join(mask_output_dir, f"{base}_mask.png")
        mask.save(mp)

        image_paths.append(ip)
        mask_paths.append(mp)

    print(f"✅ Máscaras generadas: {len(mask_paths)}")
    if image_paths:
        print(f"📦 Ejemplo:\n  IMG: {image_paths[0]}\n  MSK: {mask_paths[0]}")
    return image_paths, mask_paths

# Si ya tienes image_paths/mask_paths en memoria, comenta la siguiente línea:
try:
    image_paths, mask_paths
except NameError:
    image_paths, mask_paths = build_image_and_mask_lists()

assert len(image_paths) == len(mask_paths) and len(image_paths) > 0, "No hay datos para evaluar."
```

```python
# ---------- 2) Preprocesado estándar (letterbox 1024 + normalización + prompts) ----------
IM_SIZE = 1024
IMAGENET_MEAN = torch.tensor([0.485, 0.456, 0.406]).view(3,1,1)
IMAGENET_STD  = torch.tensor([0.229, 0.224, 0.225]).view(3,1,1)

def letterbox_rgb(pil_img, out_size=IM_SIZE):
    W, H = pil_img.size
    s = out_size / max(W, H)
    Wp, Hp = int(round(W * s)), int(round(H * s))
    # imagen: bilinear
    img_resized = pil_img.resize((Wp, Hp), resample=Image.BILINEAR)
    canvas = Image.new('RGB', (out_size, out_size), (0,0,0))
    padx = (out_size - Wp) // 2
    pady = (out_size - Hp) // 2
    canvas.paste(img_resized, (padx, pady))
    # [3,H,W] float [0,1] y copia uint8 HWC para FastSAM
    x = torch.from_numpy(np.array(canvas)).permute(2,0,1).float() / 255.0
    x_uint8_hwc = np.array(canvas, dtype=np.uint8)
    return x, x_uint8_hwc, s, (padx, pady), (Wp, Hp)

def normalize_imagenet(x):
    return (x - IMAGENET_MEAN) / IMAGENET_STD

def resize_mask_letterbox(pil_mask, s, pads, inter_shape, out_size=IM_SIZE, threshold=127):
    Wp, Hp = inter_shape
    mask_resized = pil_mask.resize((Wp, Hp), resample=Image.NEAREST).convert("L")
    canvas = Image.new("L", (out_size, out_size), 0)
    padx, pady = pads
    canvas.paste(mask_resized, (padx, pady))
    m = np.array(canvas, dtype=np.uint8)
    m_bin = (m > threshold).astype(np.uint8)
    vals = np.unique(m_bin)
    assert set(vals.tolist()).issubset({0,1}), f"Máscara no binaria: valores {vals}"
    return m_bin

def mask_to_center_and_box(mask_bin):
    ys, xs = np.where(mask_bin == 1)
    if len(xs) == 0:
        return None, None
    cx = float(xs.mean())
    cy = float(ys.mean())
    x0, x1 = int(xs.min()), int(xs.max())
    y0, y1 = int(ys.min()), int(ys.max())
    return (cx, cy), [x0, y0, x1, y1]

def project_points(points_xy, s, pads):
    points_xy = np.asarray(points_xy, dtype=np.float32)
    padx, pady = pads
    pts_prime = points_xy * s
    pts_prime[:,0] += padx
    pts_prime[:,1] += pady
    return pts_prime

def box_to_points(box_xyxy, n_samples=20, rng_seed=0):
    x0,y0,x1,y1 = box_xyxy
    rng = np.random.default_rng(rng_seed)
    xs = rng.uniform(x0, x1, size=n_samples)
    ys = rng.uniform(y0, y1, size=n_samples)
    return np.stack([xs, ys], axis=1)
```

```python
# === Dataset + saneado de prompts (redefinición compacta) ===

import numpy as np
from PIL import Image
from torch.utils.data import Dataset

# Tamaño por defecto (usa el tuyo si ya existe)
try:
    IM_SIZE
except NameError:
    IM_SIZE = 1024

# Helpers para sanear prompts
def _sanitize_points(points, imgsz=1024):
    pts = np.asarray(points, dtype=np.float32)
    if pts.ndim != 2 or pts.shape[1] != 2:
        raise ValueError(f"Puntos con forma inválida: {pts.shape}")
    pts = np.rint(pts)
    pts[:, 0] = np.clip(pts[:, 0], 0, imgsz - 1)
    pts[:, 1] = np.clip(pts[:, 1], 0, imgsz - 1)
    return pts.astype(np.int32).tolist()

def _sanitize_box_xyxy(box_xyxy, imgsz=1024):
    x0, y0, x1, y1 = [float(v) for v in box_xyxy]
    x0, x1 = min(x0, x1), max(x0, x1)
    y0, y1 = min(y0, y1), max(y0, y1)
    x0, y0, x1, y1 = np.rint([x0, y0, x1, y1]).astype(np.int32)
    x0 = np.clip(x0, 0, imgsz - 1); x1 = np.clip(x1, 0, imgsz - 1)
    y0 = np.clip(y0, 0, imgsz - 1); y1 = np.clip(y1, 0, imgsz - 1)
    if x1 == x0: x1 = min(x0 + 1, imgsz - 1)
    if y1 == y0: y1 = min(y0 + 1, imgsz - 1)
    return [int(x0), int(y0), int(x1), int(y1)]

# Comprobación de dependencias mínimas usadas dentro del dataset
required = ["letterbox_rgb","normalize_imagenet","resize_mask_letterbox",
            "mask_to_center_and_box","project_points","box_to_points"]
missing = [r for r in required if r not in globals()]
if missing:
    raise NameError(f"Faltan helpers en el entorno: {missing}. Ejecuta primero las celdas donde se definen.")

class PCBTinySAMDataset(Dataset):
    def __init__(self, image_paths, mask_paths, mode="BOX", n_multi=5, rng_seed=0):
        assert len(image_paths) == len(mask_paths)
        self.image_paths = image_paths
        self.mask_paths  = mask_paths
        self.mode = mode.upper()
        self.n_multi = n_multi
        self.rng = np.random.default_rng(rng_seed)

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        ip = self.image_paths[idx]; mp = self.mask_paths[idx]
        pil_img = Image.open(ip).convert("RGB")
        pil_msk = Image.open(mp).convert("L")
        img_w, img_h = pil_img.size

        # Letterbox + normalización + GT a 1024
        x, x_uint8_hwc, s, pads, inter_shape = letterbox_rgb(pil_img, out_size=IM_SIZE)
        x_norm = normalize_imagenet(x.clone())
        m_bin_1024 = resize_mask_letterbox(pil_msk, s, pads, inter_shape, out_size=IM_SIZE, threshold=127)

        # Prompts desde la GT ORIGINAL (espacio original)
        m_orig = (np.array(pil_msk, dtype=np.uint8) > 127).astype(np.uint8)
        center_orig, box_orig = mask_to_center_and_box(m_orig)
        if center_orig is None:
            center_orig = (img_w/2.0, img_h/2.0)
            box_orig = [img_w*0.25, img_h*0.25, img_w*0.75, img_h*0.75]

        prompts_abs = {}
        if self.mode == "POINT":
            pts = np.array([center_orig], dtype=np.float32)
            prompts_abs["points"] = project_points(pts, s, pads)  # (1,2) en 1024 letterbox

        elif self.mode == "BOX":
            x0,y0,x1,y1 = box_orig
            box_corners = np.array([[x0,y0],[x1,y1]], dtype=np.float32)
            box_1024 = project_points(box_corners, s, pads)        # (2,2)
            prompts_abs["box_xyxy"] = [float(box_1024[0,0]), float(box_1024[0,1]),
                                       float(box_1024[1,0]), float(box_1024[1,1])]

        elif self.mode == "COMBO":
            pts = np.array([center_orig], dtype=np.float32)
            prompts_abs["points"] = project_points(pts, s, pads)
            x0,y0,x1,y1 = box_orig
            box_corners = np.array([[x0,y0],[x1,y1]], dtype=np.float32)
            box_1024 = project_points(box_corners, s, pads)
            prompts_abs["box_xyxy"] = [float(box_1024[0,0]), float(box_1024[0,1]),
                                       float(box_1024[1,0]), float(box_1024[1,1])]

        elif self.mode == "MULTI":
            ys, xs = np.where(m_orig == 1)
            if len(xs) >= self.n_multi:
                idxs = self.rng.choice(len(xs), size=self.n_multi, replace=False)
                pts = np.stack([xs[idxs], ys[idxs]], axis=1).astype(np.float32)
            else:
                pts = box_to_points(box_orig, n_samples=self.n_multi, rng_seed=self.rng.integers(1<<31))
            prompts_abs["points"] = project_points(pts, s, pads)
        else:
            raise ValueError(f"Modo de prompts desconocido: {self.mode}")

        # Saneado al espacio 1024: enteros y dentro de [0, IM_SIZE-1]
        if "points" in prompts_abs:
            prompts_abs["points"] = _sanitize_points(prompts_abs["points"], imgsz=IM_SIZE)
        if "box_xyxy" in prompts_abs:
            prompts_abs["box_xyxy"] = _sanitize_box_xyxy(prompts_abs["box_xyxy"], imgsz=IM_SIZE)

        return {
            "image_path": ip,
            "mask_path": mp,
            "x_uint8_1024_hwc": x_uint8_hwc,  # para FastSAM
            "x_1024_norm": x_norm,            # por si quieres otra API
            "mask_bin_1024": m_bin_1024.astype(np.uint8),
            "prompts_abs": prompts_abs
        }

print("✅ PCBTinySAMDataset listo.")



```

```python
# ---------- 4) Métricas ----------
def compute_metrics(pred_mask_bin, gt_mask_bin):
    pred = pred_mask_bin.astype(bool)
    gt   = gt_mask_bin.astype(bool)
    tp = np.logical_and(pred, gt).sum()
    fp = np.logical_and(pred, np.logical_not(gt)).sum()
    fn = np.logical_and(np.logical_not(pred), gt).sum()
    iou = tp / (tp + fp + fn + 1e-9)
    dice = (2*tp) / (2*tp + fp + fn + 1e-9)
    prec = tp / (tp + fp + 1e-9)
    rec  = tp / (tp + fn + 1e-9)
    return dict(IoU=iou, Dice=dice, Prec=prec, Rec=rec)
```

```python
 #---------- 5) Conversión robusta de 'ann' (FastSAMPrompt) a máscara binaria 1024 ----------
def ann_to_mask_1024(ann, H=IM_SIZE, W=IM_SIZE):
    """
    'ann' puede ser:
      - lista de dicts con 'segmentation' (polígonos) o 'mask'
      - np.ndarray o torch.Tensor (H,W)
    Devolvemos una máscara binaria (0/1) 1024x1024. Si vienen varias, tomamos la de mayor área.
    """
    def poly_to_mask(polys, H, W):
        m = Image.new("L", (W, H), 0)
        draw = ImageDraw.Draw(m)
        if isinstance(polys, dict) and "counts" in polys:
            # RLE no esperado aquí; fallback a vacío
            return np.zeros((H,W), np.uint8)
        if isinstance(polys, (list, tuple)) and len(polys) > 0 and isinstance(polys[0], (list, tuple)):
            # COCO: lista de listas (cada una flatten [x1,y1,x2,y2,...])
            for p in polys:
                if len(p) >= 6:
                    xy = [(p[i], p[i+1]) for i in range(0, len(p), 2)]
                    draw.polygon(xy, outline=255, fill=255)
        return (np.array(m) > 127).astype(np.uint8)

    # caso tensor/array directo
    if isinstance(ann, torch.Tensor):
        a = ann.detach().cpu().numpy()
        if a.ndim == 2:
            return (a > 0).astype(np.uint8)
    if isinstance(ann, np.ndarray):
        if ann.ndim == 2:
            return (ann > 0).astype(np.uint8)

    # lista de anotaciones
    if isinstance(ann, (list, tuple)) and len(ann) > 0:
        cand_masks = []
        for a in ann:
            if isinstance(a, dict):
                if "mask" in a and isinstance(a["mask"], (np.ndarray, torch.Tensor)):
                    m = a["mask"]
                    if isinstance(m, torch.Tensor):
                        m = m.detach().cpu().numpy()
                    if m.ndim == 3:  # (N,H,W) -> coger la 1ª
                        m = m[0]
                    cand_masks.append((m > 0).astype(np.uint8))
                elif "segmentation" in a:
                    m = poly_to_mask(a["segmentation"], H, W)
                    cand_masks.append(m)
        if len(cand_masks) == 0:
            # si es lista de arrays directos
            for a in ann:
                if isinstance(a, (np.ndarray, torch.Tensor)):
                    m = a.detach().cpu().numpy() if isinstance(a, torch.Tensor) else a
                    if m.ndim == 2:
                        cand_masks.append((m > 0).astype(np.uint8))
        if len(cand_masks):
            # elegir la de mayor área
            areas = [cm.sum() for cm in cand_masks]
            return cand_masks[int(np.argmax(areas))]
    # fallback vacío
    return np.zeros((H,W), dtype=np.uint8)
```

```python
# ---------- 6) Inferencia con FastSAM por modalidad ----------
from fastsam import FastSAM, FastSAMPrompt  # API oficial
import numpy as np

def predict_fastsam_single(model, img_uint8_1024_hwc, mode, prompts_abs, device="cuda",
                           imgsz=1024, conf=0.4, iou=0.9):
    """
    Ejecuta FastSAM y aplica el prompt:
      - POINT: points=[[x,y],...] con labels=1
      - BOX:   box_xyxy=[x0,y0,x1,y1]
      - COMBO: filtra las máscaras de box por las que contienen el punto central; si ninguna, usa la mayor área
      - MULTI: points múltiples, labels=1
    Devuelve máscara binaria 1024x1024 (np.uint8).
    """
    # 1) Segmentación "everything" (resultados base)
    everything = model(
        img_uint8_1024_hwc,
        device=device, retina_masks=True,
        imgsz=imgsz, conf=conf, iou=iou
    )
    pp = FastSAMPrompt(img_uint8_1024_hwc, everything, device=device)

    # Helper: acepta list o np.ndarray y devuelve list[[int,int], ...]
    def _ensure_points_list(pts):
        if isinstance(pts, list):
            return [[int(x), int(y)] for x, y in pts]
        arr = np.asarray(pts)
        return arr.astype(int).tolist()

    mode = mode.upper()

    if mode == "POINT":
        pts = _ensure_points_list(prompts_abs["points"])
        labels = [1] * len(pts)
        ann = pp.point_prompt(points=pts, pointlabel=labels)
        return ann_to_mask_1024(ann)

    elif mode == "BOX":
        x0, y0, x1, y1 = [int(v) for v in prompts_abs["box_xyxy"]]
        ann = pp.box_prompt(bboxes=[[x0, y0, x1, y1]])
        return ann_to_mask_1024(ann)

    elif mode == "MULTI":
        pts = _ensure_points_list(prompts_abs["points"])
        labels = [1] * len(pts)
        ann = pp.point_prompt(points=pts, pointlabel=labels)
        return ann_to_mask_1024(ann)

    elif mode == "COMBO":
        # 1) Box prompt
        x0, y0, x1, y1 = [int(v) for v in prompts_abs["box_xyxy"]]
        ann = pp.box_prompt(bboxes=[[x0, y0, x1, y1]])

        # 2) Convertir anotaciones a máscaras individuales
        m_all = []
        if isinstance(ann, (list, tuple)):
            for a in ann:
                m_all.append(ann_to_mask_1024([a]))
        else:
            m_all.append(ann_to_mask_1024(ann))

        if not m_all:
            return np.zeros((imgsz, imgsz), np.uint8)

        # 3) Elegir la que contiene el punto central (si lo hay)
        if "points" in prompts_abs and len(prompts_abs["points"]):
            cx, cy = _ensure_points_list([prompts_abs["points"][0]])[0]
            cand = [m for m in m_all if (0 <= cy < m.shape[0]) and (0 <= cx < m.shape[1]) and (m[cy, cx] == 1)]
            if cand:
                areas = [c.sum() for c in cand]
                return cand[int(np.argmax(areas))]

        # 4) Si ninguna contiene el centro → mayor área
        areas = [m.sum() for m in m_all]
        return m_all[int(np.argmax(areas))]

    else:
        raise ValueError(f"Modo desconocido: {mode}")
```
