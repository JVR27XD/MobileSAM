```python
from torch.utils.data import Dataset, DataLoader
from torchvision.transforms import ToTensor
from PIL import Image
import matplotlib.pyplot as plt
import torch

# === 4. Dataset personalizado para PyTorch ===
class PCBDefectDataset(Dataset):
    def __init__(self, image_paths, mask_paths, transform_image=None, transform_mask=None):
        self.image_paths = image_paths
        self.mask_paths = mask_paths
        self.transform_image = transform_image or ToTensor()
        self.transform_mask = transform_mask or ToTensor()

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image = Image.open(self.image_paths[idx]).convert("RGB")
        mask = Image.open(self.mask_paths[idx]).convert("L")

        image = self.transform_image(image)
        mask = self.transform_mask(mask)

        return image, mask, self.image_paths[idx]

# === 5. Instanciar y visualizar ===
dataset = PCBDefectDataset(image_paths, mask_paths)
dataloader = DataLoader(dataset, batch_size=1, shuffle=True)

image, mask, path = next(iter(dataloader))

print("🔎 Valores únicos en la máscara:", torch.unique(mask))

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(image[0].permute(1, 2, 0))
plt.title("Imagen")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(mask[0][0], cmap='gray')
plt.title("Máscara binarizada")
plt.axis("off")
plt.show()
```

```python
import torch
from efficient_sam.efficient_sam import build_efficient_sam

device = 'cuda' if torch.cuda.is_available() else 'cpu'

# === ✅ Estas configuraciones coinciden con los pesos reales
def load_efficient_sam_vitt():
    return build_efficient_sam(
        encoder_patch_embed_dim=192,
        encoder_num_heads=3,
        checkpoint="weights/efficient_sam_vitt.pt"
    ).to(device)

def load_efficient_sam_vits():
    return build_efficient_sam(
        encoder_patch_embed_dim=384,
        encoder_num_heads=6,
        checkpoint="weights/efficient_sam_vits.pt"
    ).to(device)
# === Métricas
def print_model_stats(model, name):
    dummy = torch.randn(1, 3, 1024, 1024).to(device)

    if torch.cuda.is_available():
        torch.cuda.reset_peak_memory_stats(device)

    with torch.no_grad():
        _ = model.image_encoder(dummy)

    num_params = sum(p.numel() for p in model.parameters())
    size_MB = sum(p.element_size() * p.nelement() for p in model.parameters()) / 1024**2
    vram_MB = torch.cuda.max_memory_allocated(device) / 1024**2 if torch.cuda.is_available() else 0

    print(f"\n🔍 {name}")
    print(f"📦 Nº de parámetros: {num_params:,}")
    print(f"💾 Tamaño en memoria: {size_MB:.2f} MB")
    print(f"⚡ VRAM usada en encoding: {vram_MB:.2f} MB" if torch.cuda.is_available() else "")

# === Ejecutar
efficient_sam_vitt = load_efficient_sam_vitt()
efficient_sam_vitt.eval()
print_model_stats(efficient_sam_vitt, "EfficientSAM - ViT-Tiny")

efficient_sam_vits = load_efficient_sam_vits()
efficient_sam_vits.eval()
print_model_stats(efficient_sam_vits, "EfficientSAM - ViT-Small")


```

```python
import torch, numpy as np, time, os, matplotlib.pyplot as plt
from tqdm import tqdm
from sklearn.metrics import jaccard_score, precision_score, recall_score
from PIL import Image
from torchvision import transforms as T

plt.rcParams['image.interpolation'] = 'nearest'
device = 'cuda' if torch.cuda.is_available() else 'cpu'
try: model.eval()
except: pass

IMG_SIZE = 1024  # lado máximo que usa EfficientSAM

# --- Transform (normalización SAM, tras pad a 1024×1024) ---
transform = T.Compose([
    T.ToTensor(),
    T.Normalize(mean=[123.675/255, 116.28/255, 103.53/255],
                std=[58.395/255, 57.12/255, 57.375/255]),
])

# --- Helpers ---
def dice_coef(y_true, y_pred):
    return 2*(y_true*y_pred).sum()/(y_true.sum()+y_pred.sum()+1e-7)

def pad_to_square_1024(img_pil):
    """Resize manteniendo aspecto (lado largo→1024) + pad (derecha/abajo) hasta 1024×1024.
       Devuelve (img_1024, scale, pad_x, pad_y, (new_w,new_h))."""
    w, h = img_pil.size
    scale = IMG_SIZE / max(w, h)
    new_w, new_h = int(round(w*scale)), int(round(h*scale))
    resized = img_pil.resize((new_w, new_h), Image.BILINEAR)
    canvas = Image.new("RGB", (IMG_SIZE, IMG_SIZE))
    canvas.paste(resized, (0, 0))  # padding a la derecha/abajo
    return canvas, scale, 0, 0, (new_w, new_h)

def transform_points_to_input(xy_pts, scale, pad_x=0, pad_y=0):
    """xy_pts (N,2) en píxeles originales → píxeles del input 1024×1024 (con padding)."""
    xy = np.asarray(xy_pts, dtype=np.float32).copy()
    xy[:, 0] = xy[:, 0] * scale + pad_x
    xy[:, 1] = xy[:, 1] * scale + pad_y
    xy[:, 0] = np.clip(np.round(xy[:, 0]), 0, IMG_SIZE-1)
    xy[:, 1] = np.clip(np.round(xy[:, 1]), 0, IMG_SIZE-1)
    return xy

def clamp_xy(xy, W, H):
    xy = np.asarray(xy, dtype=np.float32).copy()
    xy[:, 0] = np.clip(np.round(xy[:, 0]), 0, W-1)
    xy[:, 1] = np.clip(np.round(xy[:, 1]), 0, H-1)
    return xy

def ensure_1024(mask_np: np.ndarray) -> np.ndarray:
    mask_np = (mask_np > 0).astype(np.uint8)
    if mask_np.shape != (IMG_SIZE, IMG_SIZE):
        mask_np = np.array(Image.fromarray(mask_np*255).resize((IMG_SIZE, IMG_SIZE), Image.NEAREST))//255
    return mask_np

def select_2d_from_pred(out_tuple: tuple) -> torch.Tensor:
    """predict_masks -> (masks, iou). Devuelve (H,W) exprimiendo [0,0,0]."""
    masks = out_tuple[0] if isinstance(out_tuple, (tuple, list)) else out_tuple
    t = masks
    while torch.is_tensor(t) and t.ndim > 2:
        t = t[0]
    assert torch.is_tensor(t) and t.ndim == 2, f"Esperaba (H,W), obtuve {type(t)} {getattr(t,'shape',None)}"
    return t

def logits_or_probs_to_binary(t: torch.Tensor, thr=0.5) -> np.ndarray:
    t = t.detach()
    if float(t.min()) < 0.0 or float(t.max()) > 1.0:  # logits
        t = t.sigmoid()
    return (t > thr).to(torch.uint8).cpu().numpy()

# --- Acumuladores ---
inference_times, vram_usages = [], []
metrics = {m: {"iou": [], "dice": [], "prec": [], "rec": []} for m in ["point", "box", "combo", "multi"]}

for i in tqdm(range(len(image_paths))):
    t0 = time.time()
    if torch.cuda.is_available():
        torch.cuda.reset_peak_memory_stats()

    img_pil = Image.open(image_paths[i]).convert("RGB")
    msk_pil = Image.open(mask_paths[i]).convert("L")
    path = image_paths[i]

    gt = (np.array(msk_pil) > 127)
    coords = np.argwhere(gt)  # (N,2) -> (y,x)
    if coords.size == 0:
        print(f"⚠ Sin anotación en: {path}")
        continue

    # Centroide y caja en coords originales
    y_c, x_c = coords.mean(axis=0).astype(int)
    x0_o, y0_o = coords[:,1].min(), coords[:,0].min()
    x1_o, y1_o = coords[:,1].max(), coords[:,0].max()
    orig_w, orig_h = img_pil.size

    # ---- PREPROCESO: letterbox a 1024 ----
    img_in, scale, pad_x, pad_y, (new_w, new_h) = pad_to_square_1024(img_pil)
    inp = transform(img_in).unsqueeze(0).to(device)
    if i == 0:
        print(f"[DEBUG] input min/max/mean: {inp.min().item():.3f} / {inp.max().item():.3f} / {inp.mean().item():.3f}")

    # ---- Prompts en el ESPACIO RESIZED (SIN padding): usa sólo 'scale' ----
    pt_xy_res  = np.array([[x_c*scale, y_c*scale]], dtype=np.float32)                              # (1,2)
    box_xy_res = np.array([[x0_o*scale, y0_o*scale], [x1_o*scale, y1_o*scale]], dtype=np.float32)  # (2,2)
    step = max(1, len(coords)//5)
    sel = coords[::step][:5] if step > 0 else np.array([[y_c, x_c]])
    if len(sel) == 0: sel = np.array([[y_c, x_c]])
    mp_xy_res  = np.stack([sel[:,1]*scale, sel[:,0]*scale], axis=1).astype(np.float32)             # (N,2)

    # clamp a (new_w,new_h)
    pt_xy_res  = clamp_xy(pt_xy_res,  new_w, new_h)
    box_xy_res = clamp_xy(box_xy_res, new_w, new_h)
    mp_xy_res  = clamp_xy(mp_xy_res,  new_w, new_h)

    # ---- Tensores 4D (B=1, Q=1, P, 2) + labels (B=1, Q=1, P) ----
    pt4    = torch.from_numpy(pt_xy_res).unsqueeze(0).unsqueeze(0).to(device)                 # (1,1,1,2)
    lbl4   = torch.tensor([[[1]]], device=device, dtype=torch.int64)                          # (1,1,1)

    box4   = torch.from_numpy(box_xy_res).unsqueeze(0).unsqueeze(0).to(device)                # (1,1,2,2)
    boxlb4 = torch.tensor([[[2, 3]]], device=device, dtype=torch.int64)                       # (1,1,2)

    combo4  = torch.cat([pt4, box4], dim=2)                                                   # (1,1,3,2)
    combol4 = torch.cat([lbl4, boxlb4], dim=2)                                                # (1,1,3)

    multi4  = torch.from_numpy(mp_xy_res).unsqueeze(0).unsqueeze(0).to(device)                # (1,1,N,2)
    multil4 = torch.tensor([[[1]*mp_xy_res.shape[0]]], device=device, dtype=torch.int64)      # (1,1,N)

    with torch.inference_mode():
        emb = model.image_encoder(inp)
        if i == 0:
            print(f"[DEBUG] emb shape={tuple(emb.shape)}, min={emb.min().item():.3f}, max={emb.max().item():.3f}, mean|abs|={emb.abs().mean().item():.3f}")

        # ⬅️ TAMAÑO DE ENTRADA = (new_h,new_w) (tras RESIZE sin pad)
        sizes = (new_h, new_w, orig_h, orig_w)  # (input_h, input_w, output_h, output_w)

        out_point = model.predict_masks(emb, pt4,    lbl4,    False, *sizes)
        out_box   = model.predict_masks(emb, box4,   boxlb4,  False, *sizes)
        out_combo = model.predict_masks(emb, combo4, combol4, False, *sizes)
        out_multi = model.predict_masks(emb, multi4, multil4, False, *sizes)

        mask_point = select_2d_from_pred(out_point)   # (orig_h, orig_w)
        mask_box   = select_2d_from_pred(out_box)
        mask_combo = select_2d_from_pred(out_combo)
        mask_multi = select_2d_from_pred(out_multi)

        if i == 0:
            for n, m in {"point":mask_point,"box":mask_box,"combo":mask_combo,"multi":mask_multi}.items():
                print(f"[DEBUG] {n} mask 2D shape: {tuple(m.shape)}, min/max/mean: {m.min().item():.3f}/{m.max().item():.3f}/{m.mean().item():.3f}")

    # --- Métricas (comparamos en 1024×1024) ---
    gt_1024 = (np.array(msk_pil.resize((IMG_SIZE, IMG_SIZE), resample=Image.NEAREST)) > 127)
    y_true = gt_1024.flatten().astype(np.uint8)

    preds = {"point": mask_point, "box": mask_box, "combo": mask_combo, "multi": mask_multi}
    for name, m in preds.items():
        y_pred = ensure_1024(logits_or_probs_to_binary(m)).flatten()
        metrics[name]["iou"].append(jaccard_score(y_true, y_pred, zero_division=0))
        metrics[name]["dice"].append(dice_coef(y_true, y_pred))
        metrics[name]["prec"].append(precision_score(y_true, y_pred, zero_division=0))
        metrics[name]["rec"].append(recall_score(y_true, y_pred, zero_division=0))

    inference_times.append(time.time() - t0)
    if torch.cuda.is_available():
        vram_usages.append(torch.cuda.max_memory_allocated() / 1024**2)

    # --- Visualización (primeras 3) ---
    if i < 3:
        # Para PLOT: usa coords con padding (se ven sobre img_in 1024×1024)
        pt_xy_in  = transform_points_to_input(np.array([[x_c, y_c]]),                            scale, pad_x, pad_y)[0].astype(int)
        box_xy_in = transform_points_to_input(np.array([[x0_o, y0_o], [x1_o, y1_o]]),            scale, pad_x, pad_y).astype(int)
        mp_xy_in  = transform_points_to_input(np.stack([sel[:,1], sel[:,0]], axis=1),            scale, pad_x, pad_y).astype(int)

        vis_bin = {k: ensure_1024(logits_or_probs_to_binary(v)) for k, v in preds.items()}
        order = [("Punto","point"), ("Caja","box"), ("Punto + Caja","combo"), ("Multipunto","multi")]
        fig, axs = plt.subplots(4, 3, figsize=(12, 14))
        for r,(title,key) in enumerate(order):
            axs[r,0].imshow(img_in)
            if key=="point":
                x,y = pt_xy_in; axs[r,0].plot(x,y,'o',markersize=6)
            elif key=="box":
                (x0,y0),(x1,y1) = box_xy_in
                rect = plt.Rectangle((x0,y0), x1-x0, y1-y0, fill=False, linewidth=2)
                axs[r,0].add_patch(rect)
            elif key=="combo":
                x,y = pt_xy_in; axs[r,0].plot(x,y,'o',markersize=6)
                (x0,y0),(x1,y1) = box_xy_in
                rect = plt.Rectangle((x0,y0), x1-x0, y1-y0, fill=False, linewidth=2)
                axs[r,0].add_patch(rect)
            else:
                for (x,y) in mp_xy_in: axs[r,0].plot(int(x), int(y), 'o', markersize=4)
            axs[r,0].set_title(f"Imagen + {title}"); axs[r,0].axis('off')

            axs[r,1].imshow(gt_1024.astype(np.uint8), cmap='gray', vmin=0, vmax=1)
            axs[r,1].set_title("GT"); axs[r,1].axis('off')

            axs[r,2].imshow(vis_bin[key].astype(np.uint8), cmap='gray', vmin=0, vmax=1)
            axs[r,2].set_title(f"Predicción ({title})"); axs[r,2].axis('off')

        fig.suptitle(os.path.basename(path), fontsize=14)
        plt.tight_layout(); plt.show()

# --- Promedios ---
print("\n📊 Métricas promedio EfficientSAM (por modalidad):")
for name in ["point","box","combo","multi"]:
    iou_m, iou_s = np.mean(metrics[name]["iou"]),  np.std(metrics[name]["iou"])
    dic_m, dic_s = np.mean(metrics[name]["dice"]), np.std(metrics[name]["dice"])
    pre_m, pre_s = np.mean(metrics[name]["prec"]), np.std(metrics[name]["prec"])
    rec_m, rec_s = np.mean(metrics[name]["rec"]),  np.std(metrics[name]["rec"])
    print(f"\n— {name.upper()}")
    print(f"IoU: {iou_m:.4f} ± {iou_s:.4f}")
    print(f"Dice: {dic_m:.4f} ± {dic_s:.4f}")
    print(f"Precisión: {pre_m:.4f} ± {pre_s:.4f}")
    print(f"Recall: {rec_m:.4f} ± {rec_s:.4f}")

print(f"\n⏱ Tiempo promedio: {np.mean(inference_times):.3f} ± {np.std(inference_times):.3f} s")
if vram_usages:
    print(f"💾 VRAM promedio: {np.mean(vram_usages):.2f} ± {np.std(vram_usages):.2f} MB")
```
