```python
# ==== BLOQUE DE RESCATE TinySAM (robusto y autocontenible) ====
import sys, os, time, inspect, importlib
import numpy as np
import torch
import torch.nn.functional as F
from PIL import Image
from tqdm import tqdm
import torchvision.transforms as T

# Añadir la ruta del repo
for p in ["/content/TinySAM", "."]:
    if os.path.isdir(p) and p not in sys.path:
        sys.path.append(p)

# -----------------------------
# 1) Encontrar el "builder"
# -----------------------------
def _find_tinysam_builder():
    """
    Busca en tinysam.* una función tipo build_* que construya TinySAM.
    Devuelve (callable, modulo_que_lo_contiene) o lanza ImportError.
    """
    importlib.import_module("tinysam")  # asegura import base
    candidate_names = [
        "build_sam_vit_tiny", "build_sam_vit_t", "build_tinysam",
        "build_tiny_sam", "build_sam_tiny_vit", "build_tiny_vit_sam"
    ]
    for mod_name in ["tinysam.build_sam", "tinysam.modeling.build_sam", "tinysam"]:
        try:
            m = importlib.import_module(mod_name)
        except Exception:
            continue
        # probar candidatos explícitos
        for fn in candidate_names:
            if hasattr(m, fn) and callable(getattr(m, fn)):
                return getattr(m, fn), m
        # fallback: cualquier build* que contenga "sam"
        for name, obj in vars(m).items():
            if callable(obj) and ("build" in name.lower()) and ("sam" in name.lower()):
                return obj, m
    raise ImportError("No se encontró un constructor tipo build_* en tinysam.*")

# -----------------------------
# 2) Carga del modelo
# -----------------------------
device = "cuda" if torch.cuda.is_available() else "cpu"

def load_tinysam(ckpt_path):
    build_fn, _ = _find_tinysam_builder()
    sig = inspect.signature(build_fn)
    try:
        if "checkpoint" in sig.parameters:
            model = build_fn(checkpoint=ckpt_path)
        else:
            model = build_fn()
            state = torch.load(ckpt_path, map_location="cpu")
            if isinstance(state, dict) and "state_dict" in state: state = state["state_dict"]
            if isinstance(state, dict) and "model" in state:      state = state["model"]
            missing, unexpected = model.load_state_dict(state, strict=False)
            if missing or unexpected:
                print(f"[load_state_dict] missing={len(missing)} unexpected={len(unexpected)}")
    except TypeError:
        # Fallback genérico a load_state_dict si la firma no encaja
        model = build_fn()
        state = torch.load(ckpt_path, map_location="cpu")
        if isinstance(state, dict) and "state_dict" in state: state = state["state_dict"]
        if isinstance(state, dict) and "model" in state:      state = state["model"]
        missing, unexpected = model.load_state_dict(state, strict=False)
        if missing or unexpected:
            print(f"[load_state_dict] missing={len(missing)} unexpected={len(unexpected)}")
    model.to(device).eval()
    print(f"✅ Constructor usado: {build_fn.__module__}.{build_fn.__name__}")
    return model

# -----------------------------
# 3) Prepro imagen + API de inferencia
# -----------------------------
to_tensor_1024 = T.Compose([
    T.Resize((1024, 1024)),
    T.ToTensor(),
    T.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225]),
])

@torch.inference_mode()
def encode_image(model, pil_im):
    x = to_tensor_1024(pil_im).unsqueeze(0).to(device)
    return model.image_encoder(x)

def _prep_points(point_list):
    if len(point_list) == 0:
        pts = np.zeros((0,2), dtype=np.float32); lbl = np.zeros((0,), dtype=np.int64)
    else:
        pts = np.array(point_list, dtype=np.float32); lbl = np.ones((len(point_list),), dtype=np.int64)
    return torch.from_numpy(pts)[None,...].to(device), torch.from_numpy(lbl)[None,...].to(device)

def _prep_box(box_xyxy):
    if box_xyxy is None:
        return None  # TinySAM prefiere None en vez de tensores vacíos
    return torch.tensor(box_xyxy, dtype=torch.float32, device=device)[None,None,:]

def _get_dense_pe_from_prompt_encoder(pe):
    """
    Devuelve el positional encoding denso del prompt_encoder,
    compatible con commits distintos.
    """
    if hasattr(pe, "get_dense_pe") and callable(pe.get_dense_pe):
        return pe.get_dense_pe()
    for cand in ("dense_pe", "pe", "positional_encoding", "image_pe"):
        if hasattr(pe, cand):
            v = getattr(pe, cand)
            return v() if callable(v) else v
    raise AttributeError("No se pudo obtener 'image_pe' del prompt_encoder.")

def _scale_point_and_box_to_1024(center_xy, box_xyxy, img_w, img_h):
    sx, sy = 1024.0 / img_w, 1024.0 / img_h
    cx, cy = center_xy
    center_1024 = (int(round(cx * sx)), int(round(cy * sy)))
    if box_xyxy is None:
        box_1024 = None
    else:
        x0, y0, x1, y1 = box_xyxy
        box_1024 = (int(round(x0 * sx)), int(round(y0 * sy)),
                    int(round(x1 * sx)), int(round(y1 * sy)))
    return center_1024, box_1024

@torch.inference_mode()
def predict_mask_from_prompts(
    model,
    image_embeddings,
    point_coords=None,
    point_labels=None,
    boxes=None,
    multimask_output=False,  # si el decoder no lo soporta, se ignora
):
    # Normaliza inputs: None si no hay prompts
    has_points = point_coords is not None and point_coords.numel() > 0
    has_labels = point_labels is not None and point_labels.numel() > 0
    points_arg = (point_coords, point_labels) if (has_points and has_labels) else None
    boxes_arg  = boxes if (boxes is not None and (not torch.is_tensor(boxes) or boxes.numel() > 0)) else None

    # 1) Prompt encoder
    sparse_embeddings, dense_embeddings = model.prompt_encoder(
        points=points_arg,
        boxes=boxes_arg,
        masks=None
    )

    # 2) Dense positional encoding
    image_pe = _get_dense_pe_from_prompt_encoder(model.prompt_encoder)

    # 3) Llamada flexible al decoder (con o sin multimask_output)
    decoder = model.mask_decoder
    sig = inspect.signature(decoder.forward)
    base_kwargs = dict(
        image_embeddings=image_embeddings,
        image_pe=image_pe,
        sparse_prompt_embeddings=sparse_embeddings,
        dense_prompt_embeddings=dense_embeddings,
        multimask_output=multimask_output,
    )
    accepted = {k: v for k, v in base_kwargs.items() if k in sig.parameters}
    try:
        out = decoder(**accepted)
    except TypeError:
        accepted.pop("multimask_output", None)
        try:
            out = decoder(**accepted)
        except TypeError:
            # Último recurso: llamada posicional
            out = decoder(image_embeddings, image_pe, sparse_embeddings, dense_embeddings)

    # 4) Extraer logits y upsample a 1024x1024
    low_res_masks = out[0] if isinstance(out, (list, tuple)) else out
    masks = F.interpolate(low_res_masks, size=(1024, 1024), mode="bilinear", align_corners=False)
    return masks  # logits (1,1,1024,1024) o (1,K,1024,1024)

# Wrappers de prompts
def infer_point(model, img_emb, center_xy_1024):
    pc, pl = _prep_points([center_xy_1024])
    return predict_mask_from_prompts(model, img_emb, pc, pl, None, False)

def infer_box(model, img_emb, box_xyxy_1024):
    bx = _prep_box(box_xyxy_1024)
    return predict_mask_from_prompts(model, img_emb, None, None, bx, False)

def infer_combo(model, img_emb, center_xy_1024, box_xyxy_1024):
    pc, pl = _prep_points([center_xy_1024]); bx = _prep_box(box_xyxy_1024)
    return predict_mask_from_prompts(model, img_emb, pc, pl, bx, False)

def infer_multipoint(model, img_emb, points_xy_1024):
    pc, pl = _prep_points(points_xy_1024)
    return predict_mask_from_prompts(model, img_emb, pc, pl, None, False)

# -----------------------------
# 4) Métricas
# -----------------------------
def binarize_from_logits(logits, thr=0.5):
    # logits: (1, K, H, W) o (1, 1, H, W)
    if logits.ndim == 4 and logits.shape[1] > 1:
        logits = logits[:, :1]  # usa la 1ª propuesta; alternativa: logits = logits.max(dim=1, keepdim=True).values
    prob = torch.sigmoid(logits).detach().cpu().numpy().squeeze()  # -> (H, W)
    return (prob > thr).astype(np.uint8), prob

def iou_score(gt, pr):
    inter = (gt & pr).sum()
    union = (gt | pr).sum()
    return (inter/(union+1e-7)).item()

def dice_score(gt, pr):
    inter = (gt & pr).sum()
    return (2*inter/(gt.sum()+pr.sum()+1e-7)).item()

def precision_recall(gt, pr):
    tp = (gt & pr).sum()
    fp = ((~gt.astype(bool)) & pr.astype(bool)).sum()
    fn = (gt.astype(bool) & (~pr.astype(bool))).sum()
    prec = tp/(tp+fp+1e-7)
    rec  = tp/(tp+fn+1e-7)
    return float(prec), float(rec)

def compute_center_and_box(gt_mask):
    coords = np.argwhere(gt_mask > 0)
    if coords.size == 0: return None, None
    y0x0 = coords.min(axis=0); y1x1 = coords.max(axis=0)
    y_c, x_c = coords.mean(axis=0).astype(int)
    x0, y0 = int(y0x0[1]), int(y0x0[0]); x1, y1 = int(y1x1[1]), int(y1x1[0])
    return (x_c, y_c), (x0, y0, x1, y1)

# -----------------------------
# 5) Evaluación por dataset
# -----------------------------
def eval_tinysam_on_dataset(
    image_paths, mask_paths,
    ckpt, prompt_types=("POINT","BOX","COMBO","MULTI"),
    n_multi_points=5, threshold=0.5,
    visualize=True, save_dir=None
):
    model = load_tinysam(ckpt)
    results = {k: {"IoU": [], "Dice": [], "Prec": [], "Rec": [], "Time": [], "VRAM": []} for k in prompt_types}

    if save_dir:
        os.makedirs(save_dir, exist_ok=True)

    rng = np.random.default_rng(0)

    for idx, (pimg, pmask) in enumerate(tqdm(zip(image_paths, mask_paths), total=len(image_paths))):
        pil = Image.open(pimg).convert("RGB")
        gt  = (np.array(Image.open(pmask).convert("L")) > 127).astype(np.uint8)
        if gt.sum() == 0:
            continue

        # --- GT a 1024x1024 para métricas ---
        gt_1024 = (np.array(Image.fromarray((gt*255).astype(np.uint8)).resize((1024,1024), resample=Image.NEAREST)) > 127).astype(np.uint8)

        # --- Encode imagen (a 1024x1024) ---
        t_start = time.time()
        if torch.cuda.is_available():
            torch.cuda.reset_peak_memory_stats()
        img_emb = encode_image(model, pil)
        base_t = time.time() - t_start

        # --- Prompts desde GT original ---
        center, box_xyxy = compute_center_and_box(gt)
        if center is None:
            continue

        img_w, img_h = pil.size
        center_1024, box_1024 = _scale_point_and_box_to_1024(center, box_xyxy, img_w, img_h)

        # MULTI points
        x0,y0,x1,y1 = box_xyxy
        points_multi = []
        for _ in range(n_multi_points):
            x = int(rng.integers(x0, x1+1)); y = int(rng.integers(y0, y1+1))
            if gt[y, x] > 0: points_multi.append((x,y))
        if not points_multi:
            points_multi = [center]
        points_multi_1024 = [_scale_point_and_box_to_1024((x,y), None, img_w, img_h)[0] for (x,y) in points_multi]

        # --- Ejecutar inferencias ---
        preds_bin = {}
        def run(name, logits, t0):
            pr_bin, _ = binarize_from_logits(logits, thr=threshold)
            iou = iou_score(gt_1024, pr_bin); dice = dice_score(gt_1024, pr_bin); prec, rec = precision_recall(gt_1024, pr_bin)
            t = base_t + (time.time() - t0)
            vram = torch.cuda.max_memory_allocated()/1e6 if torch.cuda.is_available() else 0.0
            for k,v in zip(["IoU","Dice","Prec","Rec","Time","VRAM"], [iou,dice,prec,rec,t,vram]):
                results[name][k].append(v)
            # Guardar para visualización a resolución original
            pr_resized = np.array(Image.fromarray(pr_bin.astype(np.uint8)*255).resize((img_w, img_h), Image.NEAREST)) > 127
            preds_bin[name] = pr_resized

        if "POINT" in prompt_types:
            t0 = time.time(); run("POINT", infer_point(model, img_emb, center_1024), t0)
        if "BOX" in prompt_types:
            t0 = time.time(); run("BOX", infer_box(model, img_emb, box_1024), t0)
        if "COMBO" in prompt_types:
            t0 = time.time(); run("COMBO", infer_combo(model, img_emb, center_1024, box_1024), t0)
        if "MULTI" in prompt_types:
            t0 = time.time(); run("MULTI", infer_multipoint(model, img_emb, points_multi_1024), t0)

        # --- Visualización y guardado ---
        if visualize:
            rows = [
                ("POINT", center, None, None),
                ("BOX", None, box_xyxy, None),
                ("COMBO", center, box_xyxy, None),
                ("MULTI", None, None, points_multi),
            ]
            fig, axes = plt.subplots(len(rows), 3, figsize=(12, 14))
            fig.suptitle(f"TinySAM – Ejemplo {idx+1}", fontsize=16)

            for r, (name, c_, b_, pm_) in enumerate(rows):
                img_prompts = draw_prompts_on_image(pil, center=c_, box=b_, points_multi=pm_)
                overlay = overlay_mask_on_image(pil, preds_bin[name].astype(np.uint8), alpha=0.45)

                axes[r, 0].imshow(img_prompts); axes[r, 0].set_title(f"{name} – Prompts")
                axes[r, 1].imshow(gt, cmap="gray"); axes[r, 1].set_title("GT")
                axes[r, 2].imshow(overlay); axes[r, 2].set_title("Pred (overlay)")

                for c in range(3):
                    axes[r, c].axis("off")

            plt.tight_layout()
            if save_dir:
                out_name = os.path.join(save_dir, f"tinysam_example_{idx+1}.png")
                plt.savefig(out_name, dpi=160, bbox_inches="tight")
            plt.show()

    # Resumen final
    summary = {}
    for k,d in results.items():
        if not d["IoU"]:
            continue
        summary[k] = {
            "IoU": float(np.mean(d["IoU"])), "Dice": float(np.mean(d["Dice"])),
            "Prec": float(np.mean(d["Prec"])), "Rec": float(np.mean(d["Rec"])),
            "Images": len(d["IoU"]), "Time": float(np.mean(d["Time"])),
            "VRAM(MB)": float(np.mean(d["VRAM"])),
        }
    return results, summary


print("✅ TinySAM listo (bloque robusto activado)")


```

```python
def eval_tinysam_on_dataset(
    image_paths, mask_paths,
    ckpt, prompt_types=("POINT","BOX","COMBO","MULTI"),
    n_multi_points=5, threshold=0.5,
):
    model = load_tinysam(ckpt)
    results = {k: {"IoU": [], "Dice": [], "Prec": [], "Rec": [], "Time": [], "VRAM": []} for k in prompt_types}

    for pimg, pmask in tqdm(zip(image_paths, mask_paths), total=len(image_paths)):
        pil = Image.open(pimg).convert("RGB")
        gt  = (np.array(Image.open(pmask).convert("L")) > 127).astype(np.uint8)
        if gt.sum() == 0:
            continue

        # --- GT a 1024x1024 para métricas ---
        gt_1024 = (np.array(Image.fromarray((gt*255).astype(np.uint8)).resize((1024,1024), resample=Image.NEAREST)) > 127).astype(np.uint8)

        # --- Encode imagen (a 1024x1024) ---
        t_start = time.time()
        if torch.cuda.is_available():
            torch.cuda.reset_peak_memory_stats()
        img_emb = encode_image(model, pil)
        base_t = time.time() - t_start

        # --- Prompts desde GT original, luego reescalados a 1024 ---
        center, box_xyxy = compute_center_and_box(gt)
        if center is None:
            continue

        img_w, img_h = pil.size
        center_1024, box_1024 = _scale_point_and_box_to_1024(center, box_xyxy, img_w, img_h)

        # MULTI: muestreamos dentro del bbox original y luego escalamos a 1024
        x0,y0,x1,y1 = box_xyxy
        rng = np.random.default_rng(0)
        points_multi = []
        for _ in range(n_multi_points):
            x = int(rng.integers(x0, x1+1)); y = int(rng.integers(y0, y1+1))
            if gt[y, x] > 0: points_multi.append((x,y))
        if not points_multi:
            points_multi = [center]
        points_multi_1024 = [ _scale_point_and_box_to_1024((x,y), None, img_w, img_h)[0] for (x,y) in points_multi ]

        def run(name, logits, t0):
            pr_bin, _ = binarize_from_logits(logits, thr=threshold)
            iou = iou_score(gt_1024, pr_bin); dice = dice_score(gt_1024, pr_bin); prec, rec = precision_recall(gt_1024, pr_bin)
            t = base_t + (time.time() - t0)
            vram = torch.cuda.max_memory_allocated()/1e6 if torch.cuda.is_available() else 0.0
            for k,v in zip(["IoU","Dice","Prec","Rec","Time","VRAM"], [iou,dice,prec,rec,t,vram]):
                results[name][k].append(v)

        if "POINT" in prompt_types:
            t0 = time.time(); run("POINT", infer_point(model, img_emb, center_1024), t0)
        if "BOX" in prompt_types:
            t0 = time.time(); run("BOX", infer_box(model, img_emb, box_1024), t0)
        if "COMBO" in prompt_types:
            t0 = time.time(); run("COMBO", infer_combo(model, img_emb, center_1024, box_1024), t0)
        if "MULTI" in prompt_types:
            t0 = time.time(); run("MULTI", infer_multipoint(model, img_emb, points_multi_1024), t0)

    # Resumen
    summary = {}
    for k,d in results.items():
        if not d["IoU"]:
            continue
        summary[k] = {
            "IoU": float(np.mean(d["IoU"])), "Dice": float(np.mean(d["Dice"])),
            "Prec": float(np.mean(d["Prec"])), "Rec": float(np.mean(d["Rec"])),
            "Images": len(d["IoU"]), "Time": float(np.mean(d["Time"])),
            "VRAM(MB)": float(np.mean(d["VRAM"])),
        }
    return results, summary
```

# Generación de máscaras binarias a partir de anotaciones LabelMe
Este bloque procesa anotaciones en formato LabelMe (archivos .json) para convertirlas en máscaras binarias alineadas con las imágenes originales. El procedimiento consiste en:

Lectura de anotaciones JSON: se abre cada archivo de anotación y se extrae el nombre y dimensiones de la imagen original, así como la lista de polígonos que representan las regiones de interés (defectos).

Reescalado de coordenadas: cada polígono se ajusta desde el sistema de coordenadas de la anotación (imageWidth, imageHeight) al tamaño real de la imagen (img_w, img_h).
Para garantizar que todos los puntos estén dentro de los límites de la imagen, se aplica un clamp de coordenadas.

Creación de máscara vacía: se genera una imagen en escala de grises (modo L) con fondo negro (valor 0) que servirá de lienzo para dibujar los polígonos.

Dibujo de regiones:

Para polígonos válidos (≥3 puntos) se rellena el área con valor blanco (255), representando la clase "defecto".

Para anotaciones en forma de rectángulo, se convierten las esquinas en un polígono de 4 puntos antes de dibujarlas.

Otros tipos de shape son ignorados.

Guardado de la máscara generada: se almacena en la carpeta generated_masks con el mismo nombre base de la imagen, añadiendo el sufijo _mask.png.

Registro de rutas: se guardan en listas (image_paths, mask_paths) las rutas absolutas de cada imagen y su máscara correspondiente, listas para ser usadas en entrenamiento o evaluación.

Resultado: un conjunto de máscaras binarias perfectamente alineadas con las imágenes originales, representando las regiones anotadas como objetos (valor 255) sobre fondo (valor 0).



```python
import os, json
from PIL import Image, ImageDraw

def scale_and_clamp_polygon(points, ann_w, ann_h, img_w, img_h):
    sx = img_w / float(ann_w)
    sy = img_h / float(ann_h)
    out = []
    for (x, y) in points:
        xi = int(round(max(0, min(img_w - 1, x * sx))))
        yi = int(round(max(0, min(img_h - 1, y * sy))))
        out.append((xi, yi))
    return out

# === Rutas ===
labeled_dir = "/content/drive/MyDrive/Colab Notebooks/SolDef_AI/Labeled"
mask_output_dir = os.path.join(labeled_dir, "generated_masks")
os.makedirs(mask_output_dir, exist_ok=True)

image_paths, mask_paths = [], []

json_files = sorted([f for f in os.listdir(labeled_dir) if f.lower().endswith(".json")])
print("JSONs encontrados:", len(json_files))

for jf in json_files:
    jp = os.path.join(labeled_dir, jf)
    try:
        with open(jp, "r") as f:
            data = json.load(f)
    except Exception as e:
        print(f"⚠️ No se pudo leer {jp}: {e}")
        continue

    image_filename = data.get("imagePath")
    if not image_filename:
        print(f"⚠️ {jp} sin imagePath, se salta.")
        continue

    ip = os.path.join(labeled_dir, image_filename)
    if not os.path.exists(ip):
        print(f"⚠️ Imagen no existe: {ip}, se salta.")
        continue

    # Tamaños
    img = Image.open(ip).convert("RGB")
    img_w, img_h = img.size
    ann_w = data.get("imageWidth", img_w)
    ann_h = data.get("imageHeight", img_h)

    # Máscara vacía (8-bit). Usamos 0/255 por compatibilidad.
    mask = Image.new("L", (img_w, img_h), 0)
    draw = ImageDraw.Draw(mask)

    n_polys = 0
    for shape in data.get("shapes", []):
        pts = shape.get("points", [])
        if not pts:
            continue
        # opcional: filtrar solo polígonos
        if shape.get("shape_type", "polygon") not in ("polygon", "polyline", "linestrip"):
            # si vienen rectángulos como "rectangle", puedes convertirlos a 4 puntos
            if shape.get("shape_type") == "rectangle" and len(pts) == 2:
                (x0, y0), (x1, y1) = pts
                pts = [(x0, y0), (x1, y0), (x1, y1), (x0, y1)]
            else:
                continue

        scaled = scale_and_clamp_polygon(pts, ann_w, ann_h, img_w, img_h)
        if len(scaled) >= 3:
            # fill=255 para objeto, outline opcional
            draw.polygon(scaled, outline=255, fill=255)
            n_polys += 1

    if n_polys == 0:
        print(f"ℹ️ Sin polígonos válidos en {jp}, se salta.")
        continue

    # Guardar máscara (misma base que imagen + sufijo)
    base = os.path.splitext(os.path.basename(image_filename))[0]
    mp = os.path.join(mask_output_dir, f"{base}_mask.png")
    mask.save(mp)

    image_paths.append(ip)
    mask_paths.append(mp)

print(f"✅ Máscaras generadas: {len(mask_paths)}")
print(f"📦 Ejemplo:\n  IMG: {image_paths[0] if image_paths else '—'}\n  MSK: {mask_paths[0] if mask_paths else '—'}")
```

```python
# === Helpers visuales + función visualize_tinysam_examples (autocontenida) ===
import os, numpy as np, cv2
import matplotlib.pyplot as plt
from PIL import Image

# ---- helpers de dibujo (visibles y bonitos) ----
def _halo_circle(img, center, r, color_fg, thickness=-1, halo=2):
    x, y = map(int, center)
    if halo > 0:
        cv2.circle(img, (x, y), r+halo, (255, 255, 255), thickness if thickness>0 else -1, lineType=cv2.LINE_AA)
    cv2.circle(img, (x, y), r, color_fg, thickness, lineType=cv2.LINE_AA)

def _crosshair(img, center, size=10, color=(0, 0, 0), thick=2):
    x, y = map(int, center)
    cv2.line(img, (x-size, y), (x+size, y), color, thick, lineType=cv2.LINE_AA)
    cv2.line(img, (x, y-size), (x, y+size), color, thick, lineType=cv2.LINE_AA)

def _label_box(img, text, org, fg=(255,255,255), bg=(0,0,0)):
    x, y = map(int, org)
    font = cv2.FONT_HERSHEY_SIMPLEX
    scale = 0.5; thick = 1; pad = 4
    (tw, th), _ = cv2.getTextSize(text, font, scale, thick)
    cv2.rectangle(img, (x, y-th-2*pad), (x+tw+2*pad, y), bg, -1)
    cv2.putText(img, text, (x+pad, y-pad), font, scale, fg, thick, cv2.LINE_AA)

def _filled_box_overlay(img, box, color=(0,255,0), alpha=0.20, edge=3):
    x0, y0, x1, y1 = map(int, box)
    x0 = max(0, min(x0, img.shape[1]-1)); x1 = max(0, min(x1, img.shape[1]-1))
    y0 = max(0, min(y0, img.shape[0]-1)); y1 = max(0, min(y1, img.shape[0]-1))
    overlay = img.copy()
    cv2.rectangle(overlay, (x0, y0), (x1, y1), color, -1)
    img[:] = cv2.addWeighted(overlay, alpha, img, 1-alpha, 0)
    cv2.rectangle(img, (x0, y0), (x1, y1), (255,255,255), edge+2, cv2.LINE_AA)
    cv2.rectangle(img, (x0, y0), (x1, y1), color, edge, cv2.LINE_AA)

def draw_prompts_on_image(pil_img, center=None, box=None, points_multi=None):
    img = np.array(pil_img).copy()  # RGB
    H, W = img.shape[:2]
    if box is not None:
        _filled_box_overlay(img, box, color=(0,200,0), alpha=0.20, edge=3)
        x0, y0, x1, y1 = map(int, box)
        _label_box(img, "BOX", (max(0, x0), max(18, y0)))
    if center is not None:
        cx, cy = map(int, center)
        cx = np.clip(cx, 0, W-1); cy = np.clip(cy, 0, H-1)
        _halo_circle(img, (cx, cy), r=7, color_fg=(0,128,255), thickness=-1, halo=3)
        _crosshair(img, (cx, cy), size=10, color=(0,0,0), thick=2)
        _label_box(img, "P", (min(W-20, cx+10), max(18, cy-10)), fg=(255,255,255), bg=(0,128,255))
    if points_multi:
        for i, (x, y) in enumerate(points_multi, start=1):
            xi, yi = int(x), int(y)
            if 0 <= xi < W and 0 <= yi < H:
                _halo_circle(img, (xi, yi), r=5, color_fg=(220,30,30), thickness=-1, halo=2)
                _label_box(img, str(i), (min(W-10, xi+8), max(18, yi-6)),
                           fg=(255,255,255), bg=(200,30,30))
    return Image.fromarray(img)

def overlay_mask_on_image(pil_img, mask_bin, alpha=0.45):
    base = np.array(pil_img).copy()
    red = np.zeros_like(base); red[..., 0] = 255
    m = (mask_bin.astype(bool))[..., None]
    out = np.where(m, (1-alpha)*base + alpha*red, base).astype(np.uint8)
    return Image.fromarray(out)

def clean_mask(mask_bin, morph_kernel=3, keep_largest=True):
    m = (mask_bin.astype(np.uint8) * 255)
    if morph_kernel and morph_kernel > 1:
        k = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (morph_kernel, morph_kernel))
        m = cv2.morphologyEx(m, cv2.MORPH_OPEN, k, iterations=1)
        m = cv2.morphologyEx(m, cv2.MORPH_CLOSE, k, iterations=1)
    if keep_largest:
        num, labels, stats, _ = cv2.connectedComponentsWithStats((m > 127).astype(np.uint8), connectivity=8)
        if num > 1:
            idx = 1 + np.argmax(stats[1:, cv2.CC_STAT_AREA])
            m = (labels == idx).astype(np.uint8) * 255
    return (m > 127).astype(np.uint8)

# ---- comprobación: necesitamos funciones del "bloque de rescate" ya definidas ----
def _require(*names):
    missing = [n for n in names if n not in globals()]
    if missing:
        raise NameError(f"Faltan definiciones {missing}. Ejecuta antes el 'BLOQUE DE RESCATE TinySAM'.")

_require('encode_image','infer_point','infer_box','infer_combo','infer_multipoint',
         'binarize_from_logits','_scale_point_and_box_to_1024','compute_center_and_box')

# ---- función principal de visualización ----
def visualize_tinysam_examples(
    model, image_paths, mask_paths,
    n_examples=4, n_multi_points=5, thr=0.5,
    save_dir=None, clean=True
):
    os.makedirs(save_dir, exist_ok=True) if save_dir else None
    rng = np.random.default_rng(0)
    model.eval()

    for idx in range(min(n_examples, len(image_paths))):
        pil = Image.open(image_paths[idx]).convert("RGB")
        gt  = (np.array(Image.open(mask_paths[idx]).convert("L")) > 127).astype(np.uint8)
        H, W = gt.shape

        center, box_xyxy = compute_center_and_box(gt)
        if center is None or box_xyxy is None:
            continue

        # multi-puntos dentro del bbox
        x0,y0,x1,y1 = box_xyxy
        points_multi = []
        for _ in range(n_multi_points):
            x = int(rng.integers(x0, x1+1)); y = int(rng.integers(y0, y1+1))
            if 0 <= y < H and 0 <= x < W and gt[y, x] > 0:
                points_multi.append((x, y))
        if not points_multi:
            points_multi = [center]

        # encode e inferir (prompts en 1024)
        img_emb = encode_image(model, pil)
        center_1024, box_1024 = _scale_point_and_box_to_1024(center, box_xyxy, W, H)
        points_multi_1024 = [_scale_point_and_box_to_1024((x,y), None, W, H)[0] for (x,y) in points_multi]

        logits_point = infer_point(model, img_emb, center_1024)
        logits_box   = infer_box(model, img_emb, box_1024)
        logits_combo = infer_combo(model, img_emb, center_1024, box_1024)
        logits_multi = infer_multipoint(model, img_emb, points_multi_1024)

        pred_point = binarize_from_logits(logits_point, thr=thr)[0]
        pred_box   = binarize_from_logits(logits_box,   thr=thr)[0]
        pred_combo = binarize_from_logits(logits_combo, thr=thr)[0]
        pred_multi = binarize_from_logits(logits_multi, thr=thr)[0]

        # volver a tamaño original
        def _up(mask):
            return np.array(Image.fromarray(mask.astype(np.uint8)*255).resize((W, H), Image.NEAREST)) > 127

        pred_point, pred_box, pred_combo, pred_multi = map(_up, [pred_point, pred_box, pred_combo, pred_multi])

        if clean:
            pred_point = clean_mask(pred_point, morph_kernel=3, keep_largest=True)
            pred_box   = clean_mask(pred_box,   morph_kernel=3, keep_largest=True)
            pred_combo = clean_mask(pred_combo, morph_kernel=3, keep_largest=True)
            pred_multi = clean_mask(pred_multi, morph_kernel=3, keep_largest=True)

        rows = [
            ("POINT", center, None, None, pred_point),
            ("BOX", None, box_xyxy, None, pred_box),
            ("COMBO", center, box_xyxy, None, pred_combo),
            ("MULTI", None, None, points_multi, pred_multi),
        ]

        fig, axes = plt.subplots(4, 3, figsize=(12, 14))
        fig.suptitle(f"TinySAM – Ejemplo {idx+1}", fontsize=16)

        for r, (name, c_, b_, pm_, pred_) in enumerate(rows):
            img_prompts = draw_prompts_on_image(pil, center=c_, box=b_, points_multi=pm_)
            overlay = overlay_mask_on_image(pil, pred_.astype(np.uint8), alpha=0.45)
            axes[r, 0].imshow(img_prompts); axes[r, 0].set_title(f"{name} – Prompts")
            axes[r, 1].imshow(gt, cmap="gray"); axes[r, 1].set_title("GT")
            axes[r, 2].imshow(overlay); axes[r, 2].set_title("Pred (overlay)")
            for c in range(3): axes[r, c].axis("off")

        plt.tight_layout()
        if save_dir:
            out_name = os.path.join(save_dir, f"tinysam_example_{idx+1}.png")
            plt.savefig(out_name, dpi=160, bbox_inches="tight")
        plt.show()


```
